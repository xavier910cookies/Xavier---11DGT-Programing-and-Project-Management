Dictionaries are super flexible in the data/object types that they can hold

```python
d = {'k1':123,'k2':[0,1,2,3],'k3':{'inside key':100}}
# k1 - int, k2 - list, k3 - another dictionary inside the dictionary
```

```python
d['k2']
```

```python
d['k3']
```

If I want to add  an item within a list in a dictionary, then it goes somthing like this

```python
d['k2'][2]
```
